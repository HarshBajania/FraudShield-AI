# Actual model evidence
Run `python -m scripts.train` first. Model selection uses policy data; test is reported only after selection. SHAP explains the base estimator before calibration.

```python
from pathlib import Path
import sys, json
ROOT = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
import pandas as pd
from IPython.display import display, Image
```

```python
metadata = json.loads((ROOT / 'models/metadata.json').read_text())
display(pd.DataFrame(metadata['model_comparison']))
display(pd.Series(metadata['test_metrics']))
display(metadata['test_uncertainty'])
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1,2,figsize=(10,4))
for axis, key, label in zip(ax, ['pr','roc'], ['Recall / precision', 'False positive rate / true positive rate']):
    curve = pd.DataFrame(metadata['curves'][key])
    axis.plot(curve.x, curve.y, color='#8f3446')
    axis.set_title(label)
plt.show()
```
