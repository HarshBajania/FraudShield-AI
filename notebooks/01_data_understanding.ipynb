```python
df_credit.groupby("IsFraud")["Amount"].agg(
    ["count", "mean", "median", "min", "max", "std"]
)
```

```python
fraud_by_type = (
    df_credit.groupby("TransactionType")["IsFraud"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

fraud_by_type
```

```python
fraud_by_location = (
    df_credit.groupby("Location")["IsFraud"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

fraud_by_location
```
